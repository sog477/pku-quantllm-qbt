## 自定义因子
使用过去数据形成高分看多的 (T,N) 面板。下面示例是十日反转；报价预测和含费用交易评价分别执行。

```python
import qbt
d = qbt.load()
m = qbt.build_market(d)
ctx = qbt.Context(d, m)
u = ctx.universe(min_amt=5e7, min_listed=60, not_st=True)
score = -ctx.ret(10)
daily, scorecard = qbt.evaluate_factor(d, score, u.mask)
display(scorecard)
r = qbt.run(d, ctx.long(u.top(score,100),gross=.98), start="2025-01-02", end=str(d.dates[-1]))
display(qbt.report(r))
```
